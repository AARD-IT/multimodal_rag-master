# Multimodal RAG (Retrieval-Augmented Generation)


## Problem Statement
<details>
<summary><strong>Problem Statement (click to expand)</strong></summary>

Organizations increasingly rely on documents such as reports, manuals, textbooks, research papers, and scanned PDFs that contain a mix of text, images, charts, and diagrams. Traditional question-answering systems work only on plain text and fail to understand or reason over visual content embedded inside documents.

Key challenges include:
- Visual information such as charts, maps, and diagrams is ignored
- Scanned or image-heavy PDFs are not searchable
- Users must manually navigate pages to find answers
- Text and images are processed in isolation, losing context

This project solves these issues by implementing a **Multimodal Retrieval-Augmented Generation (RAG) system** that understands both **textual and visual content** inside PDF documents.

The system enables natural language question answering by:
- Extracting text and images from PDFs
- Generating joint embeddings using CLIP
- Retrieving the most relevant text passages and images
- Using a large language model to generate grounded answers

This solution is designed as a **standalone, production-ready Multimodal RAG pipeline** suitable for enterprise document intelligence, research analysis, education, and knowledge-base search.

</details>


## Architecture


<details>
<summary><strong>System Architecture & Workflow (click to expand)</strong></summary>

```text
┌───────────────────────────────────────────────┐
│                 USER INTERFACE                │
│         Streamlit Web Application UI          │
│   - Upload PDF / Run Internal Demo            │
└───────────────────────────────────────────────┘
                     │
                     ▼
┌───────────────────────────────────────────────┐
│              PDF INGESTION LAYER              │
│   - Accepts PDF documents                     │
│   - Stores file bytes temporarily             │
└───────────────────────────────────────────────┘
                     │
                     ▼
┌───────────────────────────────────────────────┐
│           DOCUMENT PARSING (PyMuPDF)          │
│   - Extract page-wise text                    │
│   - Extract embedded images                   │
│   - Track page numbers & metadata             │
└───────────────────────────────────────────────┘
                     │
                     ▼
┌───────────────────────────────────────────────┐
│          MULTIMODAL EMBEDDING LAYER           │
│                                               │
│  TEXT EMBEDDINGS                              │
│   - Chunk text using Recursive Splitter       │
│   - Generate embeddings via CLIP (text)       │
│                                               │
│  IMAGE EMBEDDINGS                             │
│   - Convert images to RGB                     │
│   - Generate embeddings via CLIP (image)      │
└───────────────────────────────────────────────┘
                     │
                     ▼
┌───────────────────────────────────────────────┐
│             VECTOR STORE (FAISS)              │
│   - Stores text & image embeddings            │
│   - Maintains page & modality metadata        │
│   - Enables semantic similarity search        │
└───────────────────────────────────────────────┘
                     │
                     ▼
┌───────────────────────────────────────────────┐
│             RETRIEVAL MECHANISM               │
│   - Query embedded using CLIP (text)          │
│   - Optional page-based filtering             │
│   - Top-K multimodal context retrieval        │
└───────────────────────────────────────────────┘
                     │
                     ▼
┌───────────────────────────────────────────────┐
│        MULTIMODAL CONTEXT CONSTRUCTION        │
│   - Combine retrieved text passages           │
│   - Attach relevant images (Base64)           │
│   - Build LLM-ready multimodal message        │
└────────────────────────────────────────────── ┘
                     │
                     ▼
┌───────────────────────────────────────────────┐
│              GENERATION LAYER                 │
│   - GPT-4o via OpenRouter                     │
│   - Answers grounded in retrieved context     │
│   - Text + image-aware reasoning              │
└───────────────────────────────────────────────┘
                     │
                     ▼
┌──────────────────────────────────────────────┐
│               OUTPUT & DISPLAY               │
│   - Final answer shown to user               │
│   - Context images displayed alongside       │
│   - Page-referenced responses                │
└──────────────────────────────────────────────┘
```
</details> 

## Requirements


<details>
<summary><strong>System Requirements & Dependencies (click to expand)</strong></summary>

### 1️⃣ Models Used
- **CLIP (openai/clip-vit-base-patch32)**  
  Used to generate **joint embeddings** for both text and images, enabling multimodal similarity search.
- **GPT-4o (via OpenRouter)**  
  Used in the generation step to answer user questions grounded in retrieved text and image context.

---

### 2️⃣ Python Libraries & Purpose

- **streamlit**  
  Provides the interactive web UI for uploading PDFs, running demos, and displaying answers with images.

- **transformers (CLIPModel, CLIPProcessor)**  
  Loads and runs the CLIP model for text and image embedding generation.

- **torch**  
  Required for running CLIP inference efficiently.

- **PyMuPDF (fitz)**  
  Extracts page-wise text and embedded images from PDFs.

- **Pillow (PIL)**  
  Processes images extracted from PDFs.

- **langchain-community (FAISS)**  
  Stores multimodal embeddings and enables fast semantic retrieval.

- **langchain-core / langchain-text-splitters**  
  Handles document abstraction and chunking logic.

- **openai (OpenRouter client)**  
  Sends multimodal prompts (text + images) to GPT-4o for answer generation.

- **numpy, base64, io, re, os, pathlib**  
  Utility libraries for embedding storage, image encoding, regex-based query parsing, and file handling.

---

### 3️⃣ Infrastructure Requirements
- **Execution Environment**: Local machine or cloud VM
- **GPU**: Optional (CPU is sufficient for moderate workloads)
- **Internet Access**: Required for OpenRouter API calls
- **Secure Secrets Management**:
  - Streamlit Secrets (preferred)
  - Environment Variables (fallback)

---

### 4️⃣ Supported Input
- **Document Type**: PDF
- **Content**: Text, images, diagrams, charts inside PDFs
- **Queries**: Natural language questions, including page-specific queries

</details>
